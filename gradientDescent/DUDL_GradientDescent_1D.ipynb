# Gradient descent in 1D

```python
# function (as a function)
def fx(x):
  return 3*x**2 - 3*x + 4

# derivative function
def deriv(x):
  return 6*x - 3
```

```python
# plot the function and its derivative

# define a range for x
x = np.linspace(-2,2,2001)

# plotting
plt.plot(x,fx(x), x,deriv(x))
plt.xlim(x[[0,-1]])
plt.grid()
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend(['y','dy'])
plt.show()
```

```python
# 1) Most often in DL, the model trains for a set number of iterations, which is what we do here. But there are other ways
#    of defining how long the training lasts. Modify the code so that training ends when the derivative is smaller than 
#    some threshold, e.g., 0.1. Make sure your code is robust for negative derivatives.
# learning_rate = 0.01
# modelparams2 = np.zeros((training_epochs,2))
localmin = np.random.choice(x,1)
learning_rate = .1
training_epochs = 100

threshold = 0.000000001
modelparams = np.zeros((training_epochs,2))

grad = deriv(localmin)

print('Initial localmin value:',localmin)
print(grad)

grad_values = []
localmin_values = []

while abs(grad) >= threshold:
    grad = deriv(localmin)
    grad_values.append(grad)
    localmin = localmin - learning_rate*grad
    localmin_values.append(localmin)
    # print(localmin)
    # modelparams[i,0] = localmin[0]
    # modelparams[i,1] = grad[0]

print('Final localmin value:',localmin_values[-1])
print('Final grad value:',grad_values[-1])
# 
# 2) Does this change to the code produce a more accurate result? What if you change the stopping threshold?
#  reducing the threshold has the effect of increasing the number of iterations
# 3) Can you think of any potential problems that might arise when the stopping criterion is based on the derivative 
#    instead of a specified number of training epochs?
#  we dont know how many iterations we will have. also we dont know what a good value for the threshold is before hand, given that we are
# dealing with very small numbers for the threshold its easy to set a theshold too large since you have to decrease it by
#many orders of magnitude to increase the iterations marginally
```

```python
# plot the gradient over iterations

num_values = len(localmin_values)
x_values = np.linspace(0,np.max(num_values),num=(num_values))

fig,ax = plt.subplots()
ax.plot(x_values,localmin_values)
plt.show()
```

```python
num_values = len(localmin_values)
x_values = np.linspace(0,np.max(num_values),num=(num_values))

fig,ax = plt.subplots()
ax.plot(x_values,grad_values)
plt.show()
```

```python
# plot the gradient over iterations

fig,ax = plt.subplots(1,2,figsize=(12,4))

for i in range(2):
  ax[i].plot(modelparams[:,i],'o-')
  ax[i].set_xlabel('Iteration')
  ax[i].set_title(f'Final estimated minimum: {localmin[0]:.5f}')

ax[0].set_ylabel('Local minimum')
ax[1].set_ylabel('Derivative')

plt.show()
```
